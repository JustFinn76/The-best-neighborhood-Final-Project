```python
import pandas as pd
import matplotlib.pyplot as plt

arrests_path = r"Data/arrests.csv"
citations_path = r"Data/citations.csv"
population_path = r"Data/populations.csv"

arrests = pd.read_csv(arrests_path)
citations = pd.read_csv(citations_path)
populations = pd.read_csv(population_path)

counts = (arrests['INCIDENTNEIGHBORHOOD'].value_counts()
          .add(citations['NEIGHBORHOOD'].value_counts(), fill_value=0))
pop = populations.set_index('Neighborhood')['2020_Total_Population']
per_capita = (counts / pop).dropna().sort_values(ascending=False)

print(per_capita)

per_capita.plot(kind='barh', figsize=(8, 20))
plt.show()
```
